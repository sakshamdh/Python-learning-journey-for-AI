__Python Zip Function - Parallel Iteration__

What is zip function?
1. You can iterate through multiple iterators
2. It is an iterable (we can use for loop, we can use next() statement)
3. mostly used in data preprocessing, feature engineering efficiently as we can do parallel iteration through multiple iterates

```python
lst1 = ['Pilla', 'Nilla', 'Killa']
lst2 = ['a', 'b', 'c']
```

```python
for i in lst1:
    print(i)
```

```python
output = zip(lst1,lst2)
```

```python
output
```

```python
list(output)
# Creating tuples
```

__zip object is an iterator__

```python
next(output)
```

```python
lst1 = ['Pilla', 'Nilla', 'Killa', 4] # Takes same number of elements only , if there is extra element it gets skipped
lst2 = ['a', 'b', 'c']
lst3 = [1,2,3]
```

```python
output = zip(lst1,lst2,lst3)
for i,j,k in output:
    print(i,j,k)
```

```python
list(zip())
```

```python
dict1 = {'name': 'Saksham', 'lst_name': 'Dhiman', 'Age': '22'}
dict2 = {'name': 'Sana', 'lst_name': 'Dhiman', 'Age': '24'}
```

```python
dictionary = zip(dict1.items(),dict2.items())
```

```python
for (i,j),(i2,j2) in dictionary:
    print(i, j)
    print(i2, j2)
```
