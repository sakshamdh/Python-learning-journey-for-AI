__Python List Iterables vs Iterators__

Iterable - An iterable is an object that can be looped over.

Iterator - An iterator is an object that returns elements one at a time.

```python
# List is Iterable

lst=[1,2,3,4,5,6,7]

for i in lst:
    print(i)
```

```python
lst1 = iter(lst)
```

```python
lst1
```

```python
next(lst1)
```

```python
for i in lst1:
    print(i)
```
