__Sets__

A Set is an unordered collection data type that is iterable, mutuable, and has no duplicate elements. Python's set class represents the mathematical notion of a set. This is based on a data structure known as a __HASH TABLE__.

__HASH TABLE__

A hash table is a data structure that stores data in key–value pairs and allows very fast access.

Set doesn't support __Indexing__ neither any object is subscriptable 

__Sets__ use __()__ Brackets

```python
# Defining an empty set

set_var = set()
print(set_var)
print(type(set_var))
```

```python
set_var = {1,2,3,4,3}
```

```python
set_var
```

```python
set_var = {'Avengers', 'Ironman', 'Hitman'}
print(set_var)
type(set_var)
```

```python
# Inbuilt function in sets

set_var.add('Hulk')
```

```python
print(set_var)
```

```python
set1 = {'Hitman', 'Avengers', 'Ironman'}
set2 = {'Hitman', 'Avengers', 'Ironman', 'Hulk'}
```

```python
set2.intersection(set1)
```

```python
set2.difference(set1)
```

```python
set2
```

```python
# Difference update

set2.difference_update(set1)
```

```python
set2
```

__Dictionaries__ use __{}__ brackets

A dictionary is a collection which is unordered, changeable and indexed. In python dictionaries are written with curly brackets, and they have keys and values.

```python
dic = {}
```

```python
type(dic)
```

```python
type(dict())
```

```python
set_ex = {1,2,3,4,5}
```

```python
type(set_ex)
```

```python
# Lets create a dictionary

my_dict = {'Car1': 'Audi', 'Car2': 'BMW', 'Car3': 'Mercidies Benz'}
```

```python
type(my_dict)
```

```python
# Access the item values based on keys

my_dict['Car1']
```

```python
# We can even loop through the dictionaries keys

for x in my_dict:
    print(x)
```

```python
# We can even loop through the dictionaries values

for x in my_dict.values():
    print(x)
```

```python
# We can also check both keys and values

for x in my_dict.items():
    print(x)
```

```python
# Adding items in Dictionaries

my_dict['Car4'] = 'Fortuner'
```

```python
my_dict
```

```python
# If you overwrite any value to a key it will overappend and replace the previous value with new one

my_dict['Car1'] = 'Maruti'
```

```python
# Updated Car 1 value

my_dict
```

__Nested Dictionary__

```python
car1_model = {'Mercedes' : 1960}
car2_model = {'Audi' : 1970}
car3_model = {'Ambassador' : 1980}

car_type = {'Car1' : car1_model, 'Car2' : car2_model, 'Car3' : car3_model}
```

```python
print(car_type)
```

```python
# Accessing the items in the dictionary 

print(car_type['Car1'])
```

```python
print(car_type['Car2']['Audi'])
```

__Tuples__ use __()__ brackets

It is unmutable and we can't change single value in a tuple, complete Tuple can be changed

```python
# Create an empty Tuple

my_tuple = tuple()
```

```python
type(my_tuple)
```

```python
my_tuple = ()
```

```python
type(my_tuple)
```

```python
my_tuple = ('Saksham', 'Dhiman', 'Kumar')
```

```python
# Can't change single item in Tuple

my_tuple[0] = 'Manni'
```

```python
# Whole item can be changed

my_tuple = ('Hello', "World")
```

```python
print(my_tuple)
print(type(my_tuple))
```

```python
type(my_tuple)
```

```python
# Inbuilt Function

my_tuple.count('Hello')
```

```python
my_tuple.index('World')
```
