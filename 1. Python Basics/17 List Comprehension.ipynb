__List Comprehension__

List comprehension provide a concise way to create lists. It consists of brackets containing an expression followed by a for clause, then zero or more for or if clauses. The expression can be anything, meaning you can put in all kinds of object in lists.

```python
lst1=[]
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1
```

```python
lst_square([1,2,3,4,5,6,7,8])
```

```python
lst = [1,2,3,4,5,6,7,8]
```

```python
# List Comprehension

[i*i for i in lst]
```

```python
lst2 = [i*i for i in lst]
```

```python
print(lst2)
```

```python
# only for even nuumbers 

[i*i for i in lst if i%2==0]
```

```python
# only for odd numbers 

[i*i for i in lst if i%2!=0]
```
