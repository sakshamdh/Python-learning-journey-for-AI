__Static Methods__

A static method is a method that belongs to a class but does NOT use: self (object), cls (class)

It behaves like a normal function placed inside a class that does not depend on the object or class state.

```python
import datetime
now = datetime.datetime.now()
```

```python
now
```

```python
print(now.year)
type(now.year)
```

```python
class Car:
    base_price=100000  ## Class Variable
    def __init__(self,windows,doors,power):
        self.windows=windows
        self.doors=doors
        self.power=power
    def what_base_price(self):
        print(f"The base price is {self.base_price}")
    @classmethod
    def revise_base_price(cls,inflation):
        cls.base_price=cls.base_price+cls.base_price*inflation
    @staticmethod
    def check_year():
        if now.year==2026:
            return True
        else:
            return False
```

```python
Car.check_year()
```

```python
car1=Car(4,5,2000)
```

```python
if(car1.check_year):
    pass
else:
    Car.revise_base_price
```
