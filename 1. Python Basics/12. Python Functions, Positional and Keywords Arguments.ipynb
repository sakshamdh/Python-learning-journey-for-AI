```python
def even_odd(num):
    if num%2==0:
        print('The number is even')
    else:
        return 'The number is odd'
```

```python
# positional argument - Name
# keyword argument (having key value pair) - Age

def hello(name,age=29):
    print("My name is {} and age is {}".format(name,age))
```

```python
hello('Saksham')
```

```python
def hilo(*args,**kwargs):
    print(args)
    print(kwargs)
```

```python
hilo('Saksham','Dhiman',age=23,DOB=2004)
```

```python
lst=['Saksham','Dhiman']
dict={'age':23,'DOB':2004}
```

```python
hilo(lst,dict)
```

```python
hilo(*lst,**dict)
```

```python
lst = [1,2,3,4,5,6,7,8,9]
```

```python
def evenoddsum(lst):
    even_sum=0
    odd_sum=0
    for i in lst:
        if i%2==0:
            even_sum=even_sum+i
        else:
            odd_sum=odd_sum+i
    return even_sum,odd_sum
```

```python
evenoddsum(lst)
```
