__String Formatting in Python__

```python
print("Hello everyone")
```

```python
def greeting(name):
    return "Hello {}. Welcome to the community".format(name)
```

```python
greeting('Saksham')
```

```python
# If you're assigning a variable for place holder, ordering doesn't matter in formatting

def welcome_email(name,age):
    return "Welcome {name}. Your age is {age}".format(name=name,age=age)
```

```python
welcome_email('Saksham',23)
```
