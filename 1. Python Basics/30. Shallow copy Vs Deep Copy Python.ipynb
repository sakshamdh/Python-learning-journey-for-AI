__copy()__

__Shallow Copy__

```python
lst1 = [1,2,3,4]
lst2=lst1.copy()
```

```python
lst2[1] = 1000
```

```python
lst2,lst1
```

__Shallow Copy nested list__

```python
lst1 = [[1,2,3,4],[5,6,7,8]]
lst2=lst1.copy()  ## Shallow copy
```

```python
lst1[1][0]=100
```

```python
lst1
```

```python
lst2
```

```python
lst1.append([2,3,4,5,6])
```

```python
lst1
```

```python
lst2
```

__Deep copy__

```python
import copy
```

```python
lst1 = [1,2,3,4]
lst2=copy.deepcopy(lst1)  
```

```python
lst2
```

```python
lst2[1]=100
```

```python
lst2
```

```python
lst1
```

__In a normal list shallow == deep copy__

```python
lst1 = [[1,2,3,4],[5,6,7,8],[7,8,9]]
lst2=copy.deepcopy(lst1)  ## Deep copy
```

```python
# In this case Shallow copy isn't deep copy because it is creating a different memory location

lst2[1][0]=100
```

```python
lst2
```

```python
lst1
```
