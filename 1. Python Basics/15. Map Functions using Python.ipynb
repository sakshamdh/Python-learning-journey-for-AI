__Map Function in Python__

map() is a built-in Python function used to apply a function to every element of an iterable (like a list, tuple, etc)

```python
def even_or_odd(num):
    if num%2==0:
        return "The number {} is even".format(num)
    else:
        return "The number {} is odd".format(num)
```

```python
even_or_odd(24)
```

```python
lst = [1,2,3,4,5,6,7,89,23,54,656]
```

```python
map(even_or_odd,lst)
```

```python
list(map(even_or_odd,lst))
```
