__Lambda Function__

Anonymous function.

A function with no name.

Usually works faster than normal function.

```python
# Normal Function

def addition(a,b):
    return a+b
```

```python
addition(4,5)
```

```python
# Lambda function is only possible for single expression

addition = lambda a,b:a-b
```

```python
addition(49,2)
```

```python
def even(num):
    if num%2==0:
        return True
```

```python
even(4)
```

```python
even1 = lambda a:a%2==0
```

```python
even1(6)
```

```python
even1(9)
```

```python
def addition(x,y,z):
    return x+y+z
```

```python
addition(2,3,4)
```

```python
addition1 = lambda x,y,z:x+y+z
```

```python
addition1(2,3,4)
```
