__Filter Function in Python__

filter() is a built-in Python function used to select elements from a list (or iterable) based on a condition.

It keeps only the elements for which the condition is True.

```python
def even(num):
    if num%2==0:
        return True
```

```python
lst = [1,2,3,4,5,6,7,8,9,0]
```

```python
list(filter(even,lst))
```

```python
list(filter(lambda num:num%2==0, lst))
```

```python
list(map(lambda num:num%2==0, lst))
```
