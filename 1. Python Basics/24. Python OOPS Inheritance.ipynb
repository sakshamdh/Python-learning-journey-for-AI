__Python OOPS Inheritance__

```python
# Public variable - (normal no _)
# Protected Variables - (contains single _)
# Private Variables - (conatains double __)

# Type	        Syntax	        Accessible Outside Class?

# Public	    self.x	        ✅ Yes

# Protected	    self._x	        ⚠️ Should not

# Private	    self.__x	    ❌ No
```

```python
# All the class variables are public
# Car Blueprint 
# __init__ - initialisation constructor

class Car():
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    def drive(self):
        print("The Person drive the car")
```

```python
car = Car(4,5,'diesel')
```

```python
car.windows
```

```python
car.enginetype
```

```python
car.drive()
```

```python
class audi(Car):
    def __init__(self,windows,doors,enginetype,enableai):
        super().__init__(windows,doors,enginetype)
        self.enableai=enableai
    def selfdriving(self):
        print("Audi supports self driving")
```

```python
audiQ7=audi(5,5,'diesel',True)
```

```python
dir(audiQ7)
```

```python
audiQ7.windows
```

```python
print(audiQ7.enableai)
print(type(audiQ7.enableai))
```

```python
audiQ7.selfdriving()
```
