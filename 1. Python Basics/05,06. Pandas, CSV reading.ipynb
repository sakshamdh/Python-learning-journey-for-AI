__Pandas Tutorial__

Pandas is an open source, BSD-licensed library providing high-performance, easy-to-use data structure and data analysis tools for the Python programming language. Pandas helps you work with structured data (like tables or spreadsheets) easily and efficiently.

__Key features__

Data structures:

Series → a one-dimensional labeled array (like a column)

DataFrame → a two-dimensional table (like an Excel sheet)

Data cleaning: handle missing values, filter rows, rename columns

Data analysis: group data, calculate statistics, summarize datasets

File handling: read/write data from CSV, Excel, SQL, JSON, etc.

Time series support: work with dates and time-based data

```python
# First step is to import pandas

import pandas as pd 
import numpy as np
```

```python
# Playing with DataFrame and creating one

df = pd.DataFrame(np.arange(0,20).reshape(5,4),index = ['Row1', 'Row2', 'Row3', 'Row4', 'Row5'], columns = ['Column1', 'Column2', 'Column3', 'Column4'])
```

```python
# df.head() shows the first 5 rows of your DataFrame.

df.head()
```

```python
# Creates a new File

df.to_csv('Test1')
```

```python
# Accessing the elements 
# 1. .loc (just focusing on row indexs)
# 2. .iloc (focuses on both rows and columns indexs)

df.loc['Row1']
```

```python
# Check the type

type(df.loc['Row1'])
```

```python
# StringIO converts a string into a file-like object so libraries (like pandas) can read it as if it were a file.
# Whereas BytesIO → works with binary data (bytes)

from io import StringIO, BytesIO
```

__In Pandas, a space is treated as a real character, not ignored__

```python
data = ('col1,col2,col3\n'
            'x,y,1\n'
            'a,b,2\n'
            'c,d,3')
```

```python
type(data)
```

```python
pd.read_csv(StringIO(data))
```

```python
# Read from specific columns

df = pd.read_csv(StringIO(data), usecols = ['col1','col3'])
```

```python
df
```

```python
df.to_csv('Test1')
```

```python
data3 = ('index,a,b,c\n'
            '4,apple,bat,5.7\n'
            '8,orange,cow,10')
```

```python
# Making 0th index a first column

pd.read_csv(StringIO(data3))
```

```python
# Making 1th index a first column

pd.read_csv(StringIO(data3),index_col=0)
```

```python
pd.read_csv(StringIO(data3),index_col=2)
```

```python
data4 = ('a,b,c\n'
        '4,apple,bat\n'
        '8,orange,cow')
```

```python
pd.read_csv(StringIO(data4))
```

```python
pd.read_csv(StringIO(data4),index_col = False)
```

```python
# Combining usecols and index_col

pd.read_csv(StringIO(data4), usecols=['b', 'c'], index_col = False)
```

```python
# Quoting and Escape Characters. Very useful in NLP

data6 = 'a,b\n"hello, \\"Bob\\", nice to meet you",5' 
```

```python
pd.read_csv(StringIO(data6),escapechar='\\')
```
