__Python Assert__

Python provides the assert statement to check if a given logical expression is ture or false. Program execution proceeds only if the expression is true and raises the AssertionError when it is false. 

```python
10>10
```

```python
10>=10
```

```python
num=10
assert num>=10
```

```python
num=10
assert num>10
```

```python
try:
    num = int(input("Enter even number"))
    assert num%2==0
    print("The number is even")
except AssertionError:
    print("Please enter even number")
```
