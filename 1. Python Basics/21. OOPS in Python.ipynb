```python
class car:
    def __init__(self,window,door,enginetype):
        self.windows = window
        self.doors = door
        self.enginetype = enginetype
    def self_driving(self):
        return "This is a {} car.".format(self.enginetype)
```

```python
car3 = car(3,4,'petrol')
```

```python
car4 = car(4,5,'diesel')
car4.self_driving()
```

```python
print(car4.doors)
```

```python
print(car4.enginetype)
```
