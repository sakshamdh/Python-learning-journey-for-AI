How does Eval work

1. Parse Python expression     ----------------- “Parse” means: Break the code into parts and understand its structure before executing it.
2. Compile into a byte code
3. Evaluate the python expression
4. It will return the result

```python
var = compile("5*5","<string>","eval")
```

```python
eval(var)
```

__Globals__ are like dictionaries (key, value pairs)

```python
eval("x+50+x**2",{'x':10})
```

```python
x = 100
z = 98
eval('x+z-w',{'x':x,'z':z,'w':1000})
```

__Locals__

```python
eval('a+b+c',{},{'a':10,'b':20,'c':100})
```

```python
a=100
b=200
eval('a>b')
```
