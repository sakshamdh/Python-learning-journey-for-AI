__Python Generators Vs Iterators__

Iterator: An object that manually controls iteration using --iter__() and --next__() methods.

Generator: A simpler way to create iterators automatically using the yield keyword.

__Difference__

1. To create iterator we use iter() and to generator we use function along with yield keyword
2. Generator uses the yield keyword. It saves the local variable 
3. Generator in python helps us to write fast and compact code
4. Python iterator is much more memory efficient

```python
# List is nothing but a collection 
# Since we are able to iterate through this collection, we can call this collection as Iterable
# Iterable

lst = [1,2,3,4]
for i in lst:
    print(i)
```

__Iterator__

```python
iterable = iter(lst)
```

```python
next(iterable)
```

```python
type(iterable)
```

```python
# We can iterate

for i in iterable:
    print(i)
```

```python
try:
    print(next(iterable))
except StopIteration:
    print("The iterator is empty")
```

__Generator__

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
a
```
