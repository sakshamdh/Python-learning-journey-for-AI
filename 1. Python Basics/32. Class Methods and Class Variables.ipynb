__Class Methods and Class Variables__

__Class Variable__ - A variable shared among all objects of a class.

__Class Method__ - A method that operates on the class itself using cls.

```python
class Car:
    base_price=100000  ## Class Variable
    def __init__(self,windows,doors,power):
        self.windows=windows
        self.doors=doors
        self.power=power
    def what_base_price(self):
        print(f"The base price is {self.base_price}")
    @classmethod
    def revise_base_price(cls,inflation):
        cls.base_price=cls.base_price+cls.base_price*inflation
```

```python
car1=Car(4,5,2000)
car1.base_price
```

```python
Car.base_price
```

```python
Car.revise_base_price(0.10)
```
