__Python OOPS - Public, Private and Protected Access Modifiers__

__Public__

```python
# __init__ is a initializer method or constructor

# All the class variables are public
class Car():
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
```

```python
audi = Car(4,5,"Diesel")
```

```python
audi.windows=5
```

```python
audi.windows
```

__Protected__

```python
# All the class variables are protected
class Car():
    def __init__(self,windows,doors,enginetype):
        self._windows=windows
        self._doors=doors
        self._enginetype=enginetype
```

```python
class Truck(Car):
    def __init__(self,windows,doors,enginetype,horsepower):
        super().__init__(windows,doors,enginetype)
        self.horsepower=horsepower
```

'_doors',
 '_enginetype',
 '_windows'

  these varibales can be only accesed or modified from child class #protected

```python
truck=Truck(4,4,"Diesel",4000)
dir(truck)
```

```python
truck._doors=5
```

```python
truck._doors
```

```python
audi=Car(4,4,"Diesel")
dir(audi)
```

__Private__

```python
class Car():
    def __init__(self,windows,doors,enginetype):
        self.__windows=windows
        self.__doors=doors
        self.__enginetype=enginetype
```

'_Car__doors',

 '_Car__enginetype',

 '_Car__windows'

 these varibales can't be accesed or modified from outside of this particular class #private (that means during initializing only we can put these values after that we can not change the values)

```python
audi=Car(4,4,"Diesel")
dir(audi)
```
