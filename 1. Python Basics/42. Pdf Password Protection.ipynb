```python
from PyPDF2 import PdfReader, PdfWriter
```

```python
# open the pdf
file_pdf=PdfReader('Brimgham.pdf')
# object for pdf writer
out_pdf=PdfWriter()
```

```python
file_pdf
```

```python
len(file_pdf.pages)
```

```python
for i in range(len(file_pdf.pages)):
    page_details = file_pdf.pages[i]
    # add to the output page
    out_pdf.add_page(page_details)
```

```python
password = "Saksham@123"

out_pdf.encrypt(password)
```

wb - Write Bytes ---- for pdf

```python
with open("encryptedBrimgham.pdf","wb") as filename:
    out_pdf.write(filename)
```
