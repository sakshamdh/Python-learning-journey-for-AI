__MatplotLib Tutorial__ (Simple Visualization Library)

MatplotLib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API (lists, dicts, models) for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt, or GTK+.

Some of the major Pros of MatplotLib are:

° Generally easy to get started for simple plots

° Support for custom labels and texts 

° Great control of every element in a figure

° High-quality output in many format

° Very customizable in general

```python
import matplotlib.pyplot as plt

import numpy as np
```

```python
# Simple examples

x = np.arange(0,10)
y = np.arange(11,21)
```

```python
# Plotting using MatplotLib 
# Plt scatter

plt.scatter(x,y,c='g')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')

plt.savefig('Test.png')
```

```python
y = x*x
```

```python
# plt plot

print(plt.plot(x,y,'k--')) 
# plt.plot(x,y,'k*-')
# plt.plot(x,y,'ro')
# plt.plot(x,y,'ro--')
```

```python
# plt plot

plt.plot(x,y,'r*', linestyle ='dashed', linewidth = 5, markersize = 19)
plt.xlabel("X Axis")
plt.ylabel('Y Axis')
plt.title('2D Diagram')
```

```python
# Creating Subplots

plt.subplot(2,2,1)
plt.plot(x,y,'ro')
plt.subplot(2,2,2)
plt.plot(x,y,'g--')
plt.subplot(2,2,3)
plt.plot(x,y,'b*-')
plt.subplot(2,2,4)
plt.plot(x,y,'k')
```

```python
# compute the x and y coordinates for points on a sine curve 

x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
plt.title('Sine wave form')

# plot the points using matl=plotlib
plt.plot(x,y)
plt.show()
```

```python
# Subplot()
# Compute the x and y coordinates for points on sine and cosine curves

x = np.arange(0, 5 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Set up a subplot grid that has height 2 and width 1, and set the first such subplot as active
plt.subplot(2,1,1)

# Make the first plot
plt.plot(x,y_sin,'g--')
plt.title('Sine')

# Set the second subplot as active, and make the second plot
plt.subplot(2,1,2)
plt.plot(x,y_cos,'r--')
plt.title('Cosine')

plt.show()
```

__Bar Graph__

```python

x = [2,8,10]
y = [11,16,9]

x2 = [3,9,11]
y2 = [6,15,9]

plt.bar(x,y)
plt.bar(x2,y2, color ='g')
plt.title('Bar Graph')
plt.ylabel('Y Axis')
plt.xlabel('X Axis')

plt.show()
```

__Histogram__

```python
# bins function specifies the density 

a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27,5,5,5,61])
plt.hist(a, bins = 10)
plt.title('Histogram')
plt.show()
```

__Box Plot using Matplotlib__

```python
# Vert is used for vertical or horizontal plotting
# patch_artist is used for color in the box

data = [np.random.normal(0, std, 100) for std in range(1,4)]

# Rectangle box plot
plt.boxplot(data, vert=True, patch_artist=True)
plt.show()
```

__Pie Chart__

```python
# Data to plot
# size is the cumalitive value
# colors can be only selected from the list of pie colors

labels = 'Python', 'C++', 'Ruby', 'Java'
sizes = [213, 130, 245, 210]
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0.1, 0, -0.1, 0) # explode 1st slice

# Plot
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct = '%1.1f%%', shadow = True)
plt.axis('equal')
plt.show()
```
