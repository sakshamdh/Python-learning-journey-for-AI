__Boolean Variables__

Boolean values are two constant objects False and True.

They are used to represent truth values (other values can also be considered false or true).

In numeric contexts (for example, when used as the arguments to an arithmetic operator), they behave like the integers 0 and 1, respectively.

The build-in function bool() can be used to cast any value to a Boolean, if the value can be interpreted as a truth value

They are written as False and True, respectively.

```python
my_str = 'Saksham Dhiman'
```

```python
print(my_str.isalnum()) #check if all characters are alphanumeric
print(my_str.isalpha()) #check if all characters are alphabets
print(my_str.isdigit()) #check if all characters are digits
print(my_str.istitle()) #check if all characters are titlecase
print(my_str.isupper()) #check if all characters are uppercase
print(my_str.islower()) #check if all characters are lowercase
print(my_str.isspace()) #check if all characters are spaces
print(my_str.endswith('n')) #tests if string ends with n
print(my_str.startswith('S')) #tests if string starts with S
print(my_str.find('a')) #finds the first occurrence of a
print(my_str.rfind('a')) #finds the last occurrence of a
print(my_str.count('a')) #counts the number of occurrences of a
print(my_str.replace('a', 'A')) #replaces all occurrences of a with A
```

__Lists__

A List is a data structure that is a mutable, or changeable, ordered sequence of elements. Each element or value that is inside of a list is called an item. Just as strings are defined as characters between quotes, lists are defined by having the values between square brackets [].

__Lists__ use __[]__ Brackets
