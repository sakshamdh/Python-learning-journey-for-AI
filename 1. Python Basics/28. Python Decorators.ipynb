__Closures__

```python
def main_welcome(msg):
    def sub_welcome_class():
        print("Welcome to Saksham Learnings")
        print(msg)
        print("Learn with me")
    return sub_welcome_class()

# Here we used return because we want the inner function to run when main_welcome() is called.
```

```python
main_welcome("Saksham")
```

__Closure & initial Decorators__

```python
def main_welcome(func):
    def sub_welcome_class():
        print("Welcome to Saksham Learnings")
        print(func([1,2,3,4,5,6]))
        print("Learn with me")
    return sub_welcome_class()
```

```python
main_welcome(len)
```

__Decorator__

Here we call function within a function and then we are also passing function as a parameter

```python
def main_welcome(func):
    def sub_welcome_class():
        print("Welcome to Saksham Learnings")
        func()
        print("Learn with me")
    return sub_welcome_class()
```

```python
@main_welcome ######### shortcut way
def my_name():
    print("I am Saksham Dhiman")
```

```python
main_welcome(my_name)
```
