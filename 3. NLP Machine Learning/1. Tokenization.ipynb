```python
corpus = """Hello everyone, Welcome to Saksham Dhiman's Tutorials.
Please do watch all the codes carefully! to become expert in NLP.
"""
```

```python
print(corpus)
```

__Tokenization__

```python
# Paragraph --› sentences

from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
documents = sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

__Paragraph --› Words__

__Sentence --› Words__

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize # Here punctutation is also treated as a separate word

wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer # Here fullstop is not treated like another word, expect last word

tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordDetokenizer

tokenizer = TreebankWordDetokenizer()
```

```python
tokenizer.tokenize(corpus)
```
