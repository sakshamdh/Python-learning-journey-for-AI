__Stemming and its types__

Stemming is a process of reducing a word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).

```python
# Classification Problem
# Comments of product is a positive or negative review
# Review --› eating, eat, eaten, eats [going, goes, gone]--› go

words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

__PorterStemmer__

```python
from nltk.stem import PorterStemmer

stemming = PorterStemmer()
```

```python
for word in words:
    print(word+' ---› '+stemming.stem(word))
```

```python
stemming.stem('congratulate')
```

```python
stemming.stem("sitting")
```

__RegexpStemmer class__ (Regular expression)

NLTK has RegexStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and remove prefix and suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

__Snowball Stemmer__

'better accuracy and form than porter stemmer'

```python
from nltk.stem import SnowballStemmer, snowball

snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+" ---› "+snowballstemmer.stem(word))
```
